3. Call the API to get the list of the first *250* crypto-currencies in terms of market cap :
* Response result should be ordered by market cap in descending order.
* **vs_currency** parameter should be set to *usd*
* Store the first **10** crypto-currencies in a new python object

````python
import requests
import json

# Endpoint principal
base_url = "https://api.coingecko.com/api/v3/coins/markets"

# Paramètres
params = {
    "vs_currency": "usd",
    "order": "market_cap_desc",
    "per_page": 250,
    "page": 1
}

# Appel API
response = requests.get(base_url, params=params)

print("status code:", response)
print ("response content:", response.content)

````

````python
# Analyse de la réponse
data = response.json() # ici data est une liste de dictionnaires
# donc on peut utiliser len() directement ainsi que l'indexation

print("\nNombre total de cryptos reçues :", len(data))

# Stocker les 10 premières cryptos
top10 = data[:10]

print("\nExemple première crypto :\n", top10[0])

# Sauvegarde en JSON via content manager
with open("top10_crypto.json", "w") as f:
    json.dump(top10, f, indent=4)
    # json.dump prend le dict objet top10 et le convertit en JSON

print("\nFichier JSON 'top10_crypto.json' créé avec succès !")
````

5. Get historical OHLC prices for bitcoin over the running Year-To-Date period :
*   vs_currency parameter should be set to *usd*
*   The first element of the response lists (ex: ``` 1632355200000 ```) is called Unix Timestamp, which is a way to track time as a running total of seconds (cf. [https://www.unixtimestamp.com/](https://www.unixtimestamp.com/)). Find a way, using python, to convert this format to readable dates :
  * Hint 1 : you need to remove the 3 last zeros of each unix timestamp element for proper conversion
  * Hint 2 : you may want to use the `datetime` library in order to proceed to conversion (expected date format : year-month-day). Here is an example : [stackoverflow](https://stackoverflow.com/questions/3682748/converting-unix-timestamp-string-to-readable-date)

````python
import requests
from datetime import datetime
import json

# 1. Calcul du nombre de jours depuis le 1er janvier
today = datetime.now()
#ytd_days = (today - datetime(today.year, 1, 1)).days


# 2. Appel API OHLC
url = f"https://api.coingecko.com/api/v3/coins/bitcoin/ohlc"
params = {
    "vs_currency": "usd",
    "days": "365" # ou "30".strip() ou int(30)
}

response = requests.get(url, params=params)
ohlc_data = response.json()

````

````python
# 3. Conversion UNIX timestamp (ms → sec → date)
def ts_to_date(ts_ms):
    ts_ms = int(ts_ms)  # conversion safe
    ts_sec = ts_ms // 1000
    return datetime.utcfromtimestamp(ts_sec).strftime("%Y-%m-%d")

converted = [
    {
        "date": ts_to_date(row[0]),
        "open": row[1],
        "high": row[2],
        "low": row[3],
        "close": row[4]
    }
    for row in ohlc_data
]

# Aperçu
print(converted[:5])
````

6. Given the previous steps, create a dictionary containing the historical OHLC prices for the ten first coins over the running Year-To-Date period

````python
# Liste des 10 premières cryptos récupérées précédemment
top10_ids = [coin["id"] for coin in top10]   # ex: ["bitcoin", "ethereum", ...]

base_url = "https://api.coingecko.com/api/v3/coins/{id}/ohlc"

ohlc_dict = {}

for coin_id in top10_ids:
    url = base_url.format(id=coin_id)
    params = {
        "vs_currency": "usd",
        "days": 365   # approximation du YTD selon contraintes de l'API
    }

    response = requests.get(url, params=params)
    data = response.json()

    # gestion des erreurs pour éviter plantage du programme
    # en cas d'erreur interne du serveur API
    if "error" in data:
        print(f"Erreur pour {coin_id}: {data['error']}")
        continue

    ohlc_dict[coin_id] = data

# Aperçu du résultat
print(list(ohlc_dict.keys()))
print(len(ohlc_dict["bitcoin"]))  # nombre d'entrées OHLC pour BTC
````
