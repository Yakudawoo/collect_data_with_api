5. Create a loop to extract data from the firsts 150 characters of the series and add them to your DataFrame.

````python
base_url = "https://anapioficeandfire.com/api/characters/"
df = pd.DataFrame(columns=columns) # df vide
# les index du df iront de 0 à 149


for i in range(1, 151): # 151 exclu
    r = requests.get(base_url + str(i)) # concatenation des chaînes 
    # r= https://anapioficeandfire.com/api/characters/1

    data = r.json() # 
    df.loc[len(df)] = data 
    
    # Pour chaque colonne du DataFrame :
    # df.loc[index, colonne] = dict.get(colonne, NaN)
    
    # pandas va lire les colonnes du Df -> pour chacune
    # il cherche dict["colonne"]
    #si elle existe → il met la valeur
    #si elle n'existe pas → il met NaN


    # SI LES NOMS NE CORRESPONDENT PAS → 
    # pandas crée de NOUVELLES COLONNES
    # mais ici columns est extrait de la réponse API (sans objet)
    print(f"{i} personnage done")
````
